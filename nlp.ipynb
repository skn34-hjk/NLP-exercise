## NLP : 자연어 처리

1. 전처리 Preprocessing
     1) 정제 Cleansing
      - 특수문자, 이모티콘 등을 없애고 글자나 숫자만 남김

```python
import re
text = "Hello, 자연어 처리 전 정제 과정인 클러스터링을 해야 함 :>"
clean_text = re.sub(r"[^\w\s]", "", text)
clean_text
```

     2) 소문자 변환 Normalization
      - 대문자와 소문자를 통일. 
      - 한국어에선 해당 X

```python
clean_text = clean_text.lower()
clean_text
```

1. 전처리 Preprocessing   
     2) 토큰화 Tokenization
      - 텍스트들을 단어/문장 단위로 쪼갬 => 문장 토큰화, 단어 토큰화

```python
from nltk.tokenize import sent_tokenize, word_tokenize

text = "긴 문장을 조각조각 내는 것. 그것이 토큰화이다! 문장 토큰화도 있고 단어 토큰화도 있다"

# 문장 토큰화
sentences = sent_tokenize(text) 
print(sentences)

# 단어 토큰화
words = word_tokenize(text)
print(words)
```

1. 전처리 Preprocessing   
     2) 불용어 제거 Stopword
      - 문장에서 의미를 파악하는 데에 도움을 주지 않는 별 의미 없는 단어 (ex) a, the 등 ...) 를 제거함.

```python
# 영어의 경우 
from nltk.corpus import stopwords

text = 'But I can see us lost in the memory. August slipped away into a moment in time'
tokens = word_tokenize(text)

# 미리 정의된 영어 불용어 사전 불러오기
en_stopwords = stopwords.words('english') 

# tokens 리스트 안의 단어들 중, 불용어 사전에 없는 단어만 남김
filtered_tokens = [token for token in tokens if token.lower() not in en_stopwords]
filtered_tokens
```

```python
# 한국어의 경우
from konlpy.tag import Okt

text = '여름아 부탁해. 나의 사랑을 이루게 해 줘. 많이 힘겨웠던 나의 지난 추억 버리게.'

# 불용어 사전 불러오는 함수
def load_stopwords(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:  # 파일을 읽기 모드 r 로, 한글 안 깨지게 utf-8
        stopwords = [line.strip() for line in f]  # 파일의 각 줄 읽어와 리스트에 담음
    return stopwords

# 만들어둔 불용어 사전 함수 이용해 불러오기
ko_stopwords = load_stopwords('ko_stopwords.txt')

# 한국어 형태소 분석기 Okt 불러오기
okt = Okt()

# 문장을 형태소 단위로 쪼개기 (토큰화)
tokens = okt.morphs(text)
print(tokens)

# 불용어 제거
# 문장 쪼갠 tokens에서 단어를 하나씩 꺼내서, 불용어 아닌 것만 남김
cleaned_tokens = [token for token in tokens if token not in ko_stopwords]

print(cleaned_tokens)
```

1. 전처리 Preprocessing   
     2) 어간 추출 / 표제어 추출  Stemming / Lemmatization
      - 단어의 원형으로 통일.    
      ex) run, running, ran을 run으로 통일.

```python
from nltk.stem import WordNetLemmatizer

text = 'running run ran'
tokens = word_tokenize(text)

lemmatizer = WordNetLemmatizer()

lemmatized_list = []
for token in tokens:
    # tokens의 토큰들이 동사(pos='v')로 쓰였을 때의 원형 'run'으로 변환
    word = lemmatizer.lemmatize(token, pos='v') 
    lemmatized_list.append(word)

lemmatized_list
```

1. 벡터화 Vectorization
     1) 단어 사전 생성
      - 텍스트에 등장하는 단어들의 중복을 제거하고,     
      등장 빈도가 높은 순서대로 번호를 부여해서 단어사전 만듦

```python
from tensorflow.keras.preprocessing.text import Tokenizer

texts = ['긴 머리의 눈이 부신 그대가 좋아', '나에 대해 정말 알고 싶지 않나요', '그대를 가질 수 있다면', '드라마도 끊겠어요', '밤바다에 반짝이는 하늘에 별빛', '흔들리는 파도의 노랫소리', '그대를 느끼고 싶어']

# 사전에 없는 단어는 <OOV>로 처리하도록 설정
tokenizer = Tokenizer(oov_token='<OOV>')

# 단어사전 생성
tokenizer.fit_on_texts(texts) 
tokenizer.word_index
```

1. 벡터화 Vectorization   
     2) 정수 인코딩 Integer Encoding   
      - 실제 단어를 컴퓨터가 읽을 수 있는 숫자(인덱스 번호) 리스트로 변환하는 작업

```python
# 글자로 된 문장들을 숫자 리스트로 변환
encoded_seq = tokenizer.texts_to_sequences(texts)

encoded_seq
```

1. 벡터화 Vectorization   
     3) 패딩 Padding   
      - 문장마다 다른 단어의 길이를 똑같이 맞춰줌
      - 짧으면 0을 채우고, 길면 잘라냄

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 모든 문장의 길이를 5(maxlen=5)로 맞추기
padded_seq = pad_sequences(encoded_seq, maxlen=5, padding='post')

padded_seq
```

1. 임베딩 Embedding   
     1) 워드 임베딩 Word Emdebbing   
      - 정수인코딩값을, 의미/관계를 파악할 수 있도록 다차원 실수 벡터 공간에 배치.
      - 비슷한 의미의 단어는 가까운 위치에 놓임

```python
import torch.nn as nn

vocab_size = len(tokenizer.word_index) + 1 # 단어 사전 크기
embedding_dim = 100 # 단어의 의미를 100개의 차원으로 쪼갬

# 단어 번호를 의미가 담긴 실수 벡터로 변환하는 레이어
embedding_layer = nn.Embedding(num_embeddings=vocab_size, embedding_dim=embedding_dim, padding_idx=0)
```
