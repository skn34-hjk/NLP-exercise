```python
from collections import Counter

# 샘플 코퍼스
corpus = ["자연어 처리는 재미있다.", "자연어 처리는 컴퓨터 과학의 한 분야다.", "데이터 분석은 자연어 처리와 밀접하다."]

# 단어 단위로 코퍼스를 분리
tokenized_corpus = [sentence.split() for sentence in corpus]
flat_tokens = [word for sentence in tokenized_corpus for word in sentence]

# 단어 빈도 분석
word_freq = Counter(flat_tokens)
print("단어 빈도 분석 결과:", word_freq)
```

```python
import re

# 원시 텍스트
raw_text = "Hello!!! 자연어 처리 is FUN.   Do you agree? :)"

# 불필요한 기호 제거 (정규표현식 문법 ^:not \w : 문자,숫자,_ \s : 공백)
clean_text = re.sub(r"[^\w\s]", "", raw_text)

# 소문자 변환
clean_text = clean_text.lower()

# 공백 제거
clean_text = " ".join(clean_text.split())

print("처리된 텍스트:", clean_text)
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

# 원시 텍스트
raw_text = "자연어 처리는 데이터 과학의 한 분야다! 여러 전처리가 필요하다."

# 특수문자 제거 (!, ., ? 등)
clean_text = raw_text.translate(str.maketrans("", "", string.punctuation))

# 숫자 제거 (\d+ : 0부터 9까지의 모든 숫자들)
clean_text = re.sub(r"\\d+", "", clean_text)

# 불용어 제거
tokens = word_tokenize(clean_text)
# stop_words = set(stopwords.words("korean"))
stop_words = {'은', '는', '이', '가', '을', '를', '의', '하다'}
filtered_tokens = [word for word in tokens if word not in stop_words]

print("전처리 결과:", filtered_tokens)
```

```python
from konlpy.tag import Okt

# 텍스트
text = "자연어 처리는 정말 재미있다!"

# 형태소 분석
okt = Okt()  # 객체 생성 (= 형태소 분석기 가져옴)
morphs = okt.morphs(text)  # 최소 의미 단위 리스트 분리
pos = okt.pos(text)  # (형태소, 품사) 튜플 리스트로 품사 태깅 작업

print("형태소 분석 결과:", morphs)
print("품사 태깅 결과:", pos)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 샘플 데이터
corpus = ["자연어 처리는 재미있다.", "자연어 처리는 어렵다.", "데이터는 유용하다."]

# Bag-of-Words
vectorizer = CountVectorizer()  # BoW 백터라이저 생성
bow = vectorizer.fit_transform(corpus)  # 단어사전 학습 (생성)
print("Bag-of-Words:\\n", bow.toarray())

# TF-IDF
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(corpus)
print("TF-IDF:\\n", tfidf_matrix.toarray())
```

```python
# 샘플 텍스트
text = "자연어 처리는 재미있다. 자연어는 어렵지만 재미있다."

# 단어 빈도 계산
from collections import Counter
words = text.split()
word_count = Counter(words)

# 문장 길이 계산
sentences = text.split(". ")
sentence_lengths = [len(sentence.split()) for sentence in sentences]

print("단어 빈도:", word_count)
print("문장 길이:", sentence_lengths)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# 임의의 단어 벡터
word1 = np.array([[1, 2, 3]])
word2 = np.array([[2, 3, 4]])

# 코사인 유사도 계산
similarity = cosine_similarity(word1, word2)  # 결과가 1x1 행렬
print("코사인 유사도:", similarity[0][0])      # 행렬에서 스칼라 값만 출력
```

```python
class CNNClassifier(nn.Module):
    def __init__(self, vocab_size, embed_size, num_classes):
        super(CNNClassifier, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_size)
        self.conv = nn.Conv2d(1, 100, (3, embed_size))  # 3-gram 필터
        self.fc = nn.Linear(100, num_classes)

    def forward(self, x):
        x = self.embedding(x).unsqueeze(1)  # 채널 추가
        x = torch.relu(self.conv(x)).squeeze(3)
        x = torch.max(x, dim=2)[0]  # Max pooling
        x = self.fc(x)
        return x
```

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression

# 데이터 준비
corpus = ["자연어 처리는 재미있다", "Python은 강력하다"]
labels = [[1, 0], [0, 1]]  # 다중 레이블

# 모델 정의
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)
model = OneVsRestClassifier(LogisticRegression())
model.fit(X, labels)

# 예측
print(model.predict(vectorizer.transform(["Python이 재미있다"])))
```
