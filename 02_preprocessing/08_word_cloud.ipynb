# 워드 클라우드 (word cloud)

- 목적/용도: 문서/리뷰/뉴스에서 자주 등장하는 키워드를 한눈에 파악하는 “탐색(EDA)용 요약 시각화”
- 빈도 vs 중요도: 단순 빈도 외에도 TF-IDF(문서 내 중요도) 같은 가중치를 크기로 쓸 수 있음
- 전처리에서 특히 중요한 것
    - 불용어(조사/관사) 제거 + 표제어/어간 통일(예: kept/keep)
    - 필요하면 n-gram(예: ‘고객 이탈’, ‘머신 러닝’) 도 단어처럼 묶어서 표현
- 주의점(해석): 크기는 “많이 나온 정도”일 뿐이라 긍/부정, 문맥, 원인을 직접 말해주진 않음 → 다른 분석(감성/토픽 등)과 같이 보는 게 좋음
- 옵션 팁: 마스크 이미지 적용(형태 지정), max_words, min_freq, 색상/폰트(한글은 폰트 지정 필요)로 가독성 조절 가능

1. 텍스트 전처리
2. 단어 빈도 계산
3. 단어 크기 결정
4. 단어 배치
5. 시각화

```python
import numpy as np
import pandas as pd
from konlpy.tag import Okt      # 한국어 형태소 분석기
from collections import Counter # 빈도수 계산
from wordcloud import wordcloud # 워드클라우드 생성
import matplotlib.pyplot as plt
```

```python
spam_df = pd.read_csv('data/sms_spam.csv')

spam_df
```

```python
corpus = spam_df['CN'][0]
corpus
```

```python
okt = Okt()
nouns = okt.nouns(corpus)
nouns
```

```python
corpus = spam_df['CN']
corpus
```

```python
from tqdm.auto import tqdm

okt = Okt()
nouns = []

for corpus_temp in tqdm(corpus):  # 전체 코퍼스를 한 문장씩 순회
    nouns.extend(okt.nouns(corpus_temp))  # 명사 리스트를 추출해 nouns에 이어붙임

nouns
```

```python
word_count = Counter(nouns)  # 전체 명사 리스트를 {단어: 빈도} 형태로 집계

# 불용어 사전 로드 함수
def load_stopwords(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        stopwords = [line.strip() for line in f]  # 줄바꿈/공백 제거하여 리스트 생성 
    return stopwords

ko_stopwords = load_stopwords('ko_stopwords.txt')  # 불용어 리스트

# 불용어 제거 후 딕셔너리로 재구성
word_count = {word: count for word, count in word_count.items() if word not in ko_stopwords}
```

```python
from PIL import Image
from wordcloud import WordCloud

wordcloud = WordCloud(
    width=800,
    height=600,
    font_path='C:/Windows/Fonts/NGULIM.TTF',
    background_color='white',
    mask=np.array(Image.open('./data/dog.png'))  # 마스크 이미지 (검정) -> 배열 변환(형태 지정)
).generate_from_frequencies(word_count)  # {단어: 빈도}로 워드클라우드 생성

plt.imshow(wordcloud)
plt.axis('off')
plt.show()
```
