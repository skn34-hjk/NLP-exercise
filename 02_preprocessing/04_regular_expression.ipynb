# 정규표현식 (Regular Expression)

- 특정한 규칙을 가진 문자열을 찾기 위한 패턴
- 정규 표현식을 사용하면 대량의 텍스트 데이터에서 특정 패턴을 효율적으로 추출, 삭제, 대체 가능

### Syntax

**정규 표현식 기본 문법**

| 특수 문자 | 설명 | 예시 |
|---|---|---|
| . | 임의의 한 문자 | `a.c` : `abc`, `a1c` 등과 매치 |
| ? | 앞 문자가 0개 또는 1개 있을 때 매치 | `ab?c` : `abc`, `ac`와 매치 |
| * | 앞 문자가 0개 이상 있을 때 매치 | `ab*c` : `ac`, `abc`, `abbc` |
| + | 앞 문자가 1개 이상 있을 때 매치 | `ab+c` : `abc`, `abbc` |
| ^ | 문자열이 특정 문자로 시작할 때 매치 | `^abc` : `abcde`, `abc`와 매치 |
| $ | 문자열이 특정 문자로 끝날 때 매치 | `abc$` : `deabc`, `abc`와 매치 |
| {n} | 문자가 정확히 n번 반복될 때 매치 | `a{2}b` : `aab`와 매치 |
| {n,m} | 문자가 n번 이상 m번 이하 반복될 때 매치 | `a{2,4}b` : `aab`, `aaab`, `aaaab` |
| [ ] | 대괄호 안의 문자 중 하나와 매치 | `[abc]` : `a`, `b`, `c` |
| [^ ] | 대괄호 안의 문자 제외하고 매치 | `[^abc]` : `d`, `e`, `1` |
| \| | OR 연산자로 둘 중 하나와 매치 | `alb` : `a` 또는 `b` |

### 임의의 한 글자 .

```python
import re

reg_exp = re.compile('a.c')  # a + (임의 문자 1개) + c 형태

print(reg_exp.search('abc'))
print(reg_exp.search('abbbc'))
print(reg_exp.search('aXc'))
print(reg_exp.search('a c'))
print(reg_exp.search('ac'))
print(reg_exp.search('bc'))
```

### 수량자 * : 0개 이상

```python
reg_exp = re.compile('ab*c')  # a + b (가 0개 이상) + c 패턴

print(reg_exp.search('abc'))
print(reg_exp.search('abbbc'))
print(reg_exp.search('aXc'))
print(reg_exp.search('a c'))
print(reg_exp.search('ac'))
print(reg_exp.search('bc'))
```

### 수량자 ? : 0개 또는 1개

```python
reg_exp = re.compile('ab?c')  # a + b (가 0개 또는 1개) + c 패턴

print(reg_exp.search('abc'))
print(reg_exp.search('abbbc'))
print(reg_exp.search('aXc'))
print(reg_exp.search('a c'))
print(reg_exp.search('ac'))
print(reg_exp.search('bc'))
```

### 수량자 + : 1개 이상

```python
reg_exp = re.compile('ab+c')  # a + b (1개 이상) + c 패턴

print(reg_exp.search('abc'))
print(reg_exp.search('abbbc'))
print(reg_exp.search('aXc'))
print(reg_exp.search('a c'))
print(reg_exp.search('ac'))
print(reg_exp.search('bc'))
```

```python
reg_exp = re.compile('ab{3}c')  # a + b (n개) + c 패턴

print(reg_exp.search('abc'))
print(reg_exp.search('abbbc'))
print(reg_exp.search('aXc'))
print(reg_exp.search('a c'))
print(reg_exp.search('ac'))
print(reg_exp.search('bc'))
```

### 수랑자 {min, max} : min개 ~ max개

```python
reg_exp = re.compile('ab{1,3}c')  # a + b (n개~n개) + c 패턴

print(reg_exp.search('abc'))
print(reg_exp.search('abbbc'))
print(reg_exp.search('aXc'))
print(reg_exp.search('a c'))
print(reg_exp.search('ac'))
print(reg_exp.search('bc'))
```

### 정규 표현식에 맞는 패턴 다찾기

```python
# finditer로 패턴 매칭 위치 찾기
reg_exp = re.compile('a.c')

text = 'abcdsf;wrgos ijdokopmkqo grblmgkDF dfbklndjb erb fb sf fbdbbbwrgolm fsmlkm; kjvnkjnrapcbfm'

for temp in re.finditer(reg_exp, text):  # 매칭되는 것을 찾을 떄
    print(temp)
```

### 문자 매칭 [] : [] 안에 있는 것 중 한 글자

```python
reg_exp = re.compile('[abc]', re.IGNORECASE)  # a / b / c 중 1글자 찾기 (대소문자 무시)

print(reg_exp.search('안녕하세요, abc입니다!'))  # a
print(reg_exp.search('안녕하세요, cab입니다!'))  # c
print(reg_exp.search('안녕하세요, ABC입니다!'))  # A
print(reg_exp.search('안녕하세요, def입니다!'))
```

```python
reg_exp = re.compile('[a-zA-Z0-9]')  # 영문자(대소문자) + 숫자

print(reg_exp.findall('안녕하세요 오늘은 8월 10일 입니다. 저는 Brown 입니다.'))
```

```python
reg_exp = re.compile('[가-힣]')  # 한글

print(reg_exp.findall('안녕하세요 오늘은 8월 10일 입니다. 저는 Brown 입니다.'))
```

### 시작하는 문자열 ^

```python
reg_exp = re.compile('^who') 

print(reg_exp.search('who is who'))  # 시작이 who인지 매칭
print(reg_exp.search('is who'))

print(re.findall('who', 'who is who'))
print(re.findall('^who', 'who is who'))  # 전체에서 who로 시작하는 단어 찾기
print(re.findall('^who', 'is who'))
```

### 모듈 함수 & re 객체 메서드

#### search() : 문자열 패턴 검사

```python
reg_exp = re.compile('ab')  # ab 패턴 감지

print(reg_exp.search('abc'))
print(reg_exp.search('abbbc'))
print(reg_exp.search('aXc'))
```

### match() : 시작하는 문자열 패턴 검사

```python
reg_exp = re.compile('ab')  # ab 패턴으로 시작하는 문자열 감지

print(reg_exp.match('abc'))
print(reg_exp.match('abbbc'))
print(reg_exp.match('aXabc'))
```

### split() : 정규식 패턴으로 문자열 분할

```python
text = 'Apple Banana Orange'
split_text = re.split('[bo]', text, flags=re.IGNORECASE)  # b 또는 o(대소문자 무시)를 기준으로 분리
split_text
```

### findall() : 매칭된 결과 모두 반환

```python
text = '제 전화번호는 010-1234-5678 입니다.'

nums = re.findall('[0-9]+', text)
nums
```

```python
text = '제 전화번호는 010-1234-5678 입니다.'

nums = re.findall('[0-9]+-[0-9]+-[0-9]+', text)
nums
```

### sub() : 해당 패턴의 문자열을 대체

```python
text = 'Hello, everyone! Welcome to NLP 😻😻😻😻😻'

sub_text = re.sub('[^a-zA-Z]', '', text)  # 영문자(대소문자)에 해당하지 않는 것들은 공백 처리
sub_text
```

```python
from nltk.tokenize import RegexpTokenizer

text = "He's runner, but not a long_distance runner. His number is 1234."

tokenizer = RegexpTokenizer('[a-zA-Z0-9_]+')  # 영문자/숫자/_ 토큰 추출
# tokenizer = RegexpTokenizer(r'\w') # 동일

tokens = tokenizer.tokenize(text)
print(tokens)
```
